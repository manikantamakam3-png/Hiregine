```python
extraction_date = "2026-10-14"
resources = "Patient,Encounter,Observation,Condition"
```

```python
RESOURCES = [r.strip() for r in resources.split(",") if r.strip()]
```

```python
import requests
import json
from datetime import datetime, timezone

# --------------------------------------------------
# Configuration
# --------------------------------------------------

base_url = "https://hapi.fhir.org/baseR4"


# Fabric Lakehouse Files location
base_path = "Files/raw"

# Number of records per page
page_size = 400

# --------------------------------------------------
# Fetch and store paginated FHIR responses
# --------------------------------------------------

for resource in RESOURCES:

    print(f"\nFetching {resource}...")

    # Create resource directory
    resource_path = f"{base_path}/{resource}"
    mssparkutils.fs.mkdirs(resource_path)

    # Date-based folder

    date_path = f"{resource_path}/{extraction_date}"
    mssparkutils.fs.mkdirs(date_path)

    # Initial API URL
    next_url = f"{base_url}/{resource}"

    params = {
        "_count": page_size
    }

    page_number = 1
    total_records = 0

    # --------------------------------------------------
    # Pagination loop
    # --------------------------------------------------

    with requests.Session() as session:

        while next_url:

            print(f"Fetching {resource} - Page {page_number}")

            response = session.get(
                next_url,
                params=params,
                timeout=60
            )

            response.raise_for_status()

            # Preserve the original API response
            raw_json = response.text

            # Parse only to determine pagination and record count
            response_json = response.json()

            # Count resources in this page
            entries = response_json.get("entry", [])
            records_in_page = len(entries)

            total_records += records_in_page

            # Save each page separately to avoid overwriting
            file_path = (
                f"{date_path}/"
                f"{resource.lower()}_page_{page_number:05d}.json"
            )

            mssparkutils.fs.put(
                file_path,
                raw_json,
                True
            )

            print(
                f"Saved: {file_path} | "
                f"Records: {records_in_page}"
            )

            # Find the next page URL
            next_url = None

            for link in response_json.get("link", []):
                if link.get("relation") == "next":
                    next_url = link.get("url")
                    break

            # The next URL contains its own query parameters
            # so do not pass the original params again
            params = None

            page_number += 1

    print(
        f"Completed {resource} | "
        f"Total pages: {page_number - 1} | "
        f"Total records: {total_records}"
    )

print("\nAll FHIR resources ingested successfully.")
```
